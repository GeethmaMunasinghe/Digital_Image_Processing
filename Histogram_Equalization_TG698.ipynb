```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

**Grayscale Image**

```python
#Load the grayscale image
image=cv2.imread('jon-finlay--0_a71P4T84-unsplash.jpg',cv2.IMREAD_GRAYSCALE)

#Apply histogram equalization
equalized_image=cv2.equalizeHist(image)

#display original and equalized images
print("Original Image")
cv2_imshow(image)

print("Equalized Image")
cv2_imshow(equalized_image)

#Calculate histogram
hist_original=cv2.calcHist([image],[0],None,[256],[0,256])
hist_equalized=cv2.calcHist([equalized_image],[0],None,[256],[0,256])

#plot histogram
plt.figure(figsize=(12,4))

#original histogram
plt.subplot(1,2,1)
plt.plot(hist_original,color='gray')
plt.title('Histogram of Original image')
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")

#Equalized histogram
plt.subplot(1,2,2)
plt.plot(hist_equalized,color='black')
plt.title("Histogram of Equalized Image")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

**Color image**

```python
image=cv2.imread('konrad-koller-POykWoUsfaI-unsplash.jpg',cv2.COLOR_RGB2BGR)
cv2_imshow(image)
```

```python
B=image[:,:,0]
G=image[:,:,1]
R=image[:,:,2]

#Equalize each channel seperately
b_equi=cv2.equalizeHist(B)
g_equi=cv2.equalizeHist(G)
r_equi=cv2.equalizeHist(R)
```

```python
#Calculate histogram
B_histo=cv2.calcHist([b_equi],[0],None,[256],[0,256])
G_histo=cv2.calcHist([g_equi],[0],None,[256],[0,256])
R_histo=cv2.calcHist([r_equi],[0],None,[256],[0,256])
```

```python
#merge channels and create new image
equi_im=cv2.merge([b_equi,g_equi,r_equi])

#Visualize the equalize channels seperately
plt.imshow(b_equi)
plt.title("b_equi")
plt.show()
plt.imshow(g_equi)
plt.title("g_equi")
plt.show()
plt.imshow(r_equi)
plt.title("r_equi")
plt.show()
```

```python
#Visuaize histogram seperately
plt.subplot(2,2,1)
plt.plot(G_histo,'g')
plt.subplot(2,2,2)
plt.plot(R_histo,'r')
plt.subplot(2,2,3)
plt.plot(B_histo,'b')

#Visalize the original and equalized image seperately
cv2_imshow(image);
cv2_imshow(equi_im);
```
