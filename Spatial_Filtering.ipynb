```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
```

# **Linear Spatial Filtering**

**Smoothing Filter**

**Mean Filter**

```python
img=cv2.imread('inputimage.png',0)

#numbr of rows and columns
m,n=img.shape

#Averaging filter mask (3,3)
mask=np.ones([3,3],dtype=int)
mask=mask/9

#Convolve the 3x3 mask over the image
img_new=np.zeros([m,n])

for i in range (1,m-1):
  for j in range(1,n-1):
    temp=img[i-1,j-1]*mask[0,0]+img[i-1,j]*mask[0,1]+img[i-1,j+1]*mask[0,2]+img[i,j-1]*mask[1,0]+img[i,j]*mask[1,1]+img[i,j+1]*mask[1,2]+img[i+1,j-1]*mask[2,0]+img[i+1,j]*mask[2,1]+img[i+1,j+1]*mask[2,2]

    img_new[i,j]=temp


img_new=img_new.astype(np.uint8)
cv2.imwrite('Averaging Filtered Image.tif',img_new)
```

```python
#Display Original image
plt.imshow(img,cmap='gray')
plt.title('Original Image')
plt.axis('off')
plt.show()


#Display the blurred image
plt.imshow(img_new,cmap='gray')
plt.title('Averaging Filtered Image')
plt.axis('off')
plt.show()
```

**Gaussian Filter**

```python
image=cv2.imread('PEL-PTE-s5-4-600x800.jpg',cv2.IMREAD_COLOR)

image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)

#Apply Gaussian Filter
kernel_size=(15,15)
sigma_value=0
gaussian_blur=cv2.GaussianBlur(image_rgb,kernel_size,sigma_value)

#create figure with two subplot
fig,axes=plt.subplots(1,2,figsize=(12,6))

#display the original image
axes[0].imshow(image)
axes[0].set_title("Original Image")
axes[0].axis('off')

axes[1].imshow(image_rgb)
axes[1].set_title("Gaussian Blur")
axes[1].axis('off')

plt.show()
```

**Sharpening Filters**

```python
image=cv2.imread('PEL-PTE-s5-4-600x800.jpg',cv2.IMREAD_COLOR)

gray_image=cv2.cvtColor(image,cv2.COLOR_RGB2BGR)

#Apply Laplacian filter
laplacian=cv2.Laplacian(gray_image,cv2.CV_64F)

fig,axes=plt.subplots(1,2,figsize=(12,6))

axes[0].imshow(image)
axes[0].set_title("Original Image")
axes[0].axis('off')

axes[1].imshow(laplacian)
axes[1].set_title("Laplacian Image")
axes[1].axis('off')

plt.show()
```

**Non Linear**

```python
img_non_linear = cv2.imread("inputimage.png",0);

m, n = img_non_linear.shape

image_new = np.zeros([m,n])

for i in range(1 , m-1):
  for j in range(1, n-1):
    temp =[img_non_linear[i-1,j-1],
           img_non_linear[i-1,j],
           img_non_linear[i-1,j+1],
           img_non_linear[i,j-1],
           img_non_linear[i,j],
           img_non_linear[i,j+1],
           img_non_linear[i+1,j-1],
           img_non_linear[i+1,j],
           img_non_linear[i+1,j+1]]

    temp = sorted(temp)
    image_new[i,j] = temp[4]

image_new = image_new.astype(np.uint8)
cv2.imwrite('output_non_linear.png',image_new)
```

```python
#Display Original image
plt.imshow(img_non_linear,cmap='gray')
plt.title('Original Image')
plt.axis('off')
plt.show()


#Display the non-linear image
plt.imshow(image_new,cmap='gray')
plt.title('Non linear Image')
plt.axis('off')
plt.show()
```
