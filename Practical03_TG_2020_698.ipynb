Perform arithmatic operations on the images

Addition

```python
import cv2
from google.colab.patches import cv2_imshow  # Use cv2.imshow() in a local environment

# Read images
image1 = cv2.imread('add1.jpeg')
image2 = cv2.imread('add2.jpeg')

# Check if images are loaded
if image1 is None or image2 is None:
    print("Error: One or both images could not be loaded. Check file paths.")
else:
    # Resize image2 to match image1's shape
    image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

    # Ensure both images have the same number of channels
    if image1.shape[2] != image2.shape[2]:  # If channels don't match
        image2 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)  # Convert to 3-channel (if needed)

    # Perform addition
    result = cv2.add(image1, image2)

    # Show result
    cv2_imshow(result)
```


Substraction


```python

import cv2
from google.colab.patches import cv2_imshow  # Use cv2.imshow() in a local environment

# Read images
image1 = cv2.imread('sub1.jpg')
image2 = cv2.imread('sub2.jpg')

# Check if images are loaded correctly
if image1 is None or image2 is None:
    print("Error: One or both images could not be loaded. Check file paths.")
else:
    # Resize image2 to match image1's dimensions
    image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

    # Ensure both images have the same number of channels
    if image1.shape[2] != image2.shape[2]:  # If channels don't match
        image2 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)  # Convert to 3-channel (if needed)

    # Perform image subtraction
    result = cv2.subtract(image1, image2)

    # Show result
    cv2_imshow(result)

```

Image blending


```python

import cv2
from google.colab.patches import cv2_imshow  # Use cv2.imshow() in a local environment

# Read images
image1 = cv2.imread('add1.jpeg')
image2 = cv2.imread('add2.jpeg')

# Check if images are loaded correctly
if image1 is None or image2 is None:
    print("Error: One or both images could not be loaded. Check file paths.")
else:
    # Resize image2 to match image1's dimensions
    image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

    # Ensure both images have the same number of channels
    if image1.shape[2] != image2.shape[2]:  # If channels don't match
        image2 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)  # Convert if needed

    # Perform weighted addition
    weighted = cv2.addWeighted(image1, 0.5, image2, 0.8, -15)

    # Show result
    cv2_imshow(weighted)

```
